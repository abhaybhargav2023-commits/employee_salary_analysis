# Employee Salary Analysis

This project analyzes employee salary data using Python.

The analysis calculates:
- Total salary
- Average salary
- Highest salary
- Lowest salary

Python's built-in functions are used to perform the calculations.

```python

salaries = [45000, 55000, 38000, 72000, 60000]

print("Employee Salaries:", salaries)
```

```python
# Calculate salary statistics

total_salary = sum(salaries)
average_salary = total_salary / len(salaries)
highest_salary = max(salaries)
lowest_salary = min(salaries)

print("Total Salary:", total_salary)
print("Average Salary:", average_salary)
print("Highest Salary:", highest_salary)
print("Lowest Salary:", lowest_salary)
```

```python

print("Highest salary present in original data:", highest_salary in salaries)
print("Lowest salary present in original data:", lowest_salary in salaries)

print("\nOriginal Salary Data:", salaries)
print("Calculated Highest Salary:", highest_salary)
print("Calculated Lowest Salary:", lowest_salary)
```

```python
# Final salary analysis summary

print("===== Employee Salary Analysis =====")
print(f"Number of Employees: {len(salaries)}")
print(f"Total Salary: ₹{total_salary}")
print(f"Average Salary: ₹{average_salary:.2f}")
print(f"Highest Salary: ₹{highest_salary}")
print(f"Lowest Salary: ₹{lowest_salary}")
```
